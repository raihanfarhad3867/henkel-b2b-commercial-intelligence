# 04 • Price-volume bridge (synthetic)

Diagnostic arithmetic for eight **fictional** B2B accounts; the price/volume bridge is descriptive, not causal.


```python
from pathlib import Path
import csv
ROOT = Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
assert (ROOT/'data').exists(), 'Run from the repository root or notebooks directory'
def records(relative):
    with (ROOT/relative).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))

d=records('data/synthetic/phase4/P4_Diagnostics.csv')
price=sum(float(x['Price_Effect_EUR_CALC']) for x in d)
volume=sum(float(x['Volume_Effect_EUR_CALC']) for x in d)
change=sum(float(x['Sales_Change_EUR_CALC']) for x in d)
print('SIMULATED volume effect (€):',round(volume,2))
print('SIMULATED price effect (€):',round(price,2))
print('SIMULATED sales change (€):',round(change,2))
assert abs(price+volume-change)<0.02
```
