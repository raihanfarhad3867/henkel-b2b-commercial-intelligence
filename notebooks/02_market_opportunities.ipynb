# 02 • Market evidence and opportunity assumptions

Compare external market indicators against hypothetical market sizing. A market indicator is **not** a Henkel sales opportunity.


```python
from pathlib import Path
import csv
ROOT = Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
assert (ROOT/'data').exists(), 'Run from the repository root or notebooks directory'
def records(relative):
    with (ROOT/relative).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))

market=records('data/desk_research/P2_Market_Evidence.csv')
models=records('data/assumption_models/P2_Opportunity_Model.csv')
for x in market[:6]:
    print(x['Market'],x['Indicator'],x['Value'],x['Unit'],'| source:',x['Source_URL'])
print('Scenario count:',len(models),'— each is assumption-led and NOT a company forecast.')
assert all('SIMULATED' in x['Data_Status'].upper() for x in models)
```
