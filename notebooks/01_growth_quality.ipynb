# 01 • Growth quality (official reported data)

Uses Q3 2024–2026 Adhesive Technologies price/volume contribution data from the source-linked financial dataset. **Do not attribute the business-unit split to APC.**


```python
from pathlib import Path
import csv
ROOT = Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
assert (ROOT/'data').exists(), 'Run from the repository root or notebooks directory'
def records(relative):
    with (ROOT/relative).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))

at=records('data/official_financials/AT_Quarterly_Official_2024_2026.csv')
apc=records('data/official_financials/APC_Quarterly_Official_2024_2026.csv')
for y in ['2024','2025','2026']:
    a=next(x for x in at if x['Year']==y and x['Quarter']=='Q3')
    b=next(x for x in apc if x['Year']==y and x['Quarter']=='Q3')
    organic=float(a['AT_Organic_Growth_pct']); price=float(a['AT_Price_Contribution_pp']); vol=float(a['AT_Volume_Contribution_pp'])
    print(y, 'AT organic:', organic, 'price pp:',price,'volume pp:',vol,'APC organic:', b['APC_Organic_Growth_pct'])
    assert abs(organic-price-vol)<0.11
    assert not b['APC_Price_Contribution_pp'], 'APC price is undisclosed'
```
