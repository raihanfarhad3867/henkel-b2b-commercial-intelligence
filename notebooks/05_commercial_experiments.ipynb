The programs are simulated, matched comparisons; results are not real experiments or causal estimates about Henkel.


```python
from pathlib import Path
import csv
ROOT = Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
assert (ROOT/'data').exists(), 'Run from the repository root or notebooks directory'
def records(relative):
    with (ROOT/relative).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))

r=records('data/synthetic/phase5/P5_Effect_Estimates.csv')
for x in r:
    print(x['Program'], 'annual scenario net (€):', x['Annualized_Net_EUR_SCENARIO'], '|',x['Decision_Screen'])
assert len(r)==5
assert all('SIMULATED' in x['Data_Status'].upper() for x in r)
```
