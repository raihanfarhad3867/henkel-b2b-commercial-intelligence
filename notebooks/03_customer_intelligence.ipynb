Document public manufacturing activities; **relationship, supplier and true prospect status remain unverified**.


```python
from pathlib import Path
import csv
ROOT = Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
assert (ROOT/'data').exists(), 'Run from the repository root or notebooks directory'
def records(relative):
    with (ROOT/relative).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))

accounts=records('data/desk_research/P3_Account_Research.csv')
screen=records('data/desk_research/P3_Account_Screening.csv')
print('Public desk-research units:',len(accounts))
for x in screen:
    print(x['Research_Unit'], 'research interest:', x['Research_Interest_0to100_CALC'],'Henkel relationship:',x['Henkel_Relationship'],'Priority:',x['Commercial_Priority'])
assert all(x['Henkel_Relationship'].lower()=='unknown' for x in screen)
```
